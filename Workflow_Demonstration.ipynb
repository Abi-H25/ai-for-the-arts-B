Every step in the process of working collaboratively with the generative AI tools involved human research. Initially, this was selecting a part of the original journal, writing detailed notes and then organising that information into a prompt. 

**Step 2: Prompt Creation** 

```python
prompt = """
Generate a short, historically accurate day in the life, styled diary entry of a 19th century, wealthy businessman travelling through Nice in 1867. Please make clear his freedoms as a male traveller, his class social status and the observations he makes. Also maintain a reflective tone of a 19th century traveller with considerations that the text will be for a non-specialist audience. Discuss topics like his accommodation, his travel by boat and rail and his daily activities.
"""
print(prompt)
```
