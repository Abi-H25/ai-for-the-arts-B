**<h2>Reflections</h2>**

**This notebook contains reflective paragraphs on several parts of the design and development process of this project.**

**Human vs AI Creativity**

One of the main points for reflection from my project is the balance between human and creativity and the generative AI. While the tools of ChatGPT and DALL-E were responsible for the content in the character profiles, the overall creative process remained dependent on my human input. In terms of human input, I initially chose to work with the travel journal, conducted the historical research, thought of and designed the character profiles and continuously edited and refined the prompts used in the AI generation process to ensure they aligned with the aims of my project. My input highlighted that AI was more of a collaborative tool in the project rather than another independent creator. 

The quality of the AI generated content relied heavily on the detail from my prompts. The first versions of the AI content often felt too modern or had several historical inaccuracies that needed more input from me to improve. This demonstrated to me the principles discussed through the Lovelace and Boden theories that AI creativity relies on the tasks and constraints set out by human input. 


**Historical Accuracy and Interpretation**

One of the main challenges I experienced throughout this project was ensuring that my creative interpretation of the travel journal remained historically accurate. The Journal of a Continental Tour primarily conveyed the perspective of a wealthy male traveller, which meant that the other perspectives I set out to create were only partially represented. Due to this, some aspects of the character profiles had to be further researched or inferred from the context given in the writing.
 
Although the generative AI tools allowed me to still create the alternative character profiles and perspectives, there was a risk that they would feel oversimplified or dramatised versions of the original content in the journal. To ensure this did not happen I continuously referred to the writing and illustrations within the journal when describing the content needed for the character profiles. By doing this, I was confident that the AI generated content was suitable and accurate for the project and that is maintained the aim of using a storytelling approach to engage a wider audience. 


**Prompt development and Iterative design**

Prompt development was one of the most important parts of my projects creative process. As discussed, many of the initial AI generations were inaccurate or too modern for my project with some of the profile images depicting people with too many fingers or distorted facial features. This demonstrated to me that the AI models could not independently generate produce the accurate content without my direction.  As I prompted the generations of more content, I found that adding exact years, locations, colour schemes, weather conditions, facial expressions and clothing styles all helped considerably in the quality of the content. I realised that my own assumptions of the capabilities of AI and what input was required was not realistic and through this process came to better understand how to work alongside generative AI tools. 


**Audience Engagement and Accessibility**

The projects main aim was to make the travel journal more accessible and engaging for younger audiences and users without any specialist historical knowledge. The original Journal of a Continental Tour, while detailed, felt difficult to understand or interpret, which I felt would discourage many users from interacting with it. To combat this, I designed and used AI to generate the character profiles so that the content within the journal was broken into simpler to understand perspectives, without compromising the original work. This approach successfully created a way for users to interact with the journals content and be presented with the realities of 19th century travel in a non-overwhelming way. Additionally, from working with the AI tools for content generation and digital storytelling, I better understood how AI can be applied to enhance cultural heritage projects. 


**Ethical Concerns of Generative AI**

During this project I had several ethical concerns when using AI to reinterpret the Journal of a Continental Tour. My main concern was with the datasets used to train both the AI models of ChatGPT and DALL-E. As a user, there was no clear way to identify what training material would have contributed to my generated outputs. The training material raised questions surrounding the creative ownership of the AI generated content especially when presented as part of a reinterpretation of a cultural heritage artefact. 

Additionally, I had concerns that the generative AI would oversimplify the journals content, beyond converting the understanding of its content to fit a younger, non-specialist audience. To ensure this did not happen, I maintained strong links to the journals original content and highlight to my intended audience that the AI generations were creative interpretations. 

This process further demonstrated the importance of human oversight when working with generative AI to ensure that the AI tools were used appropriately and that ethical concerns were considered as part of the design and development process.
