**<h1>Methodology</h1>**

**This notebook contains the process of developing this project in collaboration with generative AI tools.**

**Data Selection and Source Material**

As discussed in the introduction, the primary data for this project was taken from the Journal of a Continental Tour (1867-1868) which is held in the University of Glasgow Archives (GB 247 MS Gen 13) (University of Glasgow Archives, 2024). This first section of the methodology will describe what content was taken from the journal and how it was used in the development of this project. 

The journal was treated as a dataset for recurring themes, travel motivations and social interactions to be extracted from. Specific illustrations were chosen based on their relevance to the projects aim of exploring who travelled and what were the motivations behind their journeys. Illustrations depicting modes of transport, social interaction and nuances of 19th century travel were prioritised as they provided the most insight into the lives of those on board the ship between Glasgow and France. 

The illustrations were also key to constructing the character identities, as recurring figures were adapted for the individual historical profiles. While the travel journal centres around the experiences and interactions of its author Andrew Macgeorge, the illustrations depict different and often overlook individuals, who have their own purpose and experiences throughout the journey. The depictions of a range of individuals through the hand drawn pictures allowed the project to expand beyond a single narrative. 

However, the journal did have some initial limitations. As the journal is written by Andrew Macgeorge, only one perspective was conveyed and often the perspectives of others such as the workers and women throughout the journey, although important context, were overlooked. 
To address this, the project uses the illustrations, locations and situations discussed in the journal along with broader knowledge of 19th century travel and generative AI tools to construct the alternative perspectives, with all content being thoroughly evaluated to ensure the character profiles remain true to historical reality. This approach allowed the journal to serve both as a historical record but also as a base for representing more diverse experiences of travel at the time. 


**Taking an Analytical Approach**

The journal was analysed through a process of close reading specific passages and the accompanying visual material, such as the hand drawn illustrations. This approach allowed for a more comprehensive understanding of a 19th century travel experience as the visual content often expanded upon the written entries in the journal. Rather than treating the journal as one narrative, this project examined it as a qualitative source that revealed patterns in social interactions and movements that would inform the creation of the character profiles.

From this analysis, key recurring character types and travel motivations were identified. The journals perspective was largely shaped by upper-class traveller experiences with their motivations rooted in leisure, social status and cultural exploration. However, the experiences of less visible figures, such as those working on the journey, were not explicitly documented, yet information on their perspectives could be inferred from references to the importance of their work and interactions with the upper-class travellers.

Additionally, considerations of differing experiences between genders were also highlighted as a recurring theme. Several depictions suggested the limited freedoms and independence of female travellers, which further influenced the character profiles.

The outcome of this analysis process was the transformation of the journals original narrative into four distinct character profiles and a set of unique experiences and interactions for each one. The additional contextual information found also became central for the development of the overall storytelling of the project. 


**Character Profile Construction**

As discussed, to make the journal more engaging and accessible for a young audience, four character profiles were created. This process transformed the historical accounts into easily followed “day in the life” descriptions.  The creation of the multiple perspectives helped users to better understand the social dynamics, expectations and experiences of 19th century travel. 
Below, the four historically based characters and their profiles are presented. Each one is developed based on the patterns identified in the journal. 
1.	Wealthy Businessman
- This profile represents the main perspective and social class described in the journal. The evidence within the writing of structured routes, culturally enriching experiences suggested that his motivation for travelling stemmed from personal enrichment, reinforcing social status and leisure. 
- This character is shown to be financially secure and mobile within society. His presence creates a secure baseline for understanding who had access to travel in 19th century society. 

2.	Businessman’s Wife
- The inclusion of this character profile created a gendered perspective that is less visible from initially exploring the journal but is still important overall to the experiences of 19th century travellers. While direct references to female travellers are limited, there are several depictions of social interactions and scenarios centred around the woman on board throughout the journal. 
- This character reflects the specific constraints and expectations placed upon female travellers at the time. Her role compared to her husband seems more observational and she often has restrictions on her independence. Her profile allows for an initial comparison that not all travellers had an equal experience and offers a more nuanced account of the journey. 

3.	Young Worker
- This character provides a stark contrast to the wealthier travellers as their role facilitated the journey, despite their perspective being overlooked. The profile is based largely through inferred evidence and references to transport systems and interactions with the wealthy travellers and is supported by wider context on 19th century travel norms. 
- The workers inclusion highlights the hidden labour behind elite travel and introduces a socio-economic contrast which the journal does not directly address. 

4.	Andrew Macgeorge
- The perspective of Andrew Macgrorge is not the same as the other fully detailed character profiles but rather, as the author of the journal, he serves as an overarching narrative voice. His writing shapes the way travel is depicted, and his perspective reflects the biases and priorities of his social position. 
- His inclusion allows the project the continually demonstrate the importance of the source material (the travel journal), while still acknowledging the limitations of his singular viewpoint. 

The combination of these profiles together expands the journal beyond its initial singular documentation of 19th century travel and offers a more comprehensive and accessible view. Establishing these profiles allowed for later comparison against the generated modern day traveller profiles to convey the changes in social norms and motivations. 


**Modern Day Character Comparison**

To complement the historical character profiles, two modern day traveller profiles were constructed to serve as bases for comparison to better engage the young audience. These characters are really tools, allowing users to reflect on the social and cultural changes to travel. Rather than presenting the historical figures in isolation, this approach encourages users to interact with the profiles and consider the changes for themselves. 

Modern Character 1: Young Male Traveller 

This character represents a typical travel experience in today’s society. He is in his twenties and enjoys travelling independently or with his friends. He often looks for low-cost travel options, and his motivations are centred around leisure and cultural experiences. His experience is shaped by technological advancements, from his modes of transport to how he plans and books his trip. His travel is self-directed and fits his desires, with fewer limitations compared to travel in the 19th century. 

Modern Character 2: Young Female Traveller

Similarly, this traveller represents a typical female traveller in modern society. Again, in her twenties, she has the same motives of leisure and cultural experiences but with more consideration centred around safety. Compared to the historical female travellers, she has significantly more autonomy and access to travel and makes use of modern technology as part of her experiences. Her presence allows for clear comparisons between past and present restrictions. 

Drawing the Comparisons: 

These modern characters are positioned in the project to interact with the historical profiles and to reflect on their experience with those of the 19th century profiles. 
For example, the modern-day female traveller might discuss the differences in independence and highlight modern day societies progress through the existence of her own experiences. 
The constructed interactions directly appeal to the audience by taking the historic travel journal and offering a more interactive approach to considering the advancements in modern day travel and the experiences on those traveling in the 19th century. The incorporation of generative AI tools for the profiles and storytelling improves audience accessibility and the educational takeaway for users without any speciality knowledge. 


**AI Workflow and Tools**

<h3>Part 1: Text Processing Using ChatGPT</h3>

The main textual elements of the “day in the life” character profiles as well as modern day profiles were developed using ChatGPT, which is based on Large Language Models (LLMs) (OpenAI, 2024). LLMs are trained on large datasets and use deep learning techniques, specifically neural networks to generate human-like responses. These models rely on pattern recognition to predict the most likely sequence of words based on the input it receives (Naveed et al., 2025). 

The ability of the LLM to recognise linguistic patterns made it particularly effective for this project in adapting the tone and rewriting events from different perspectives, which were central areas of this project. 

-	The first stage in generating the text for this project was completed manually. Detailed notes on key events and illustration were taken from the journal and then organised into which characters profile they would be fit. This ensured that the profiles remained in line with the recorded events from the original journal.

    By completing this stage without the use of AI, the project avoided any misinterpretation from a more automated AI analysis and ensured that all the content maintained accurate to the historical event. 

-	The second stage used ChatGPT to convert the notes into the “day in the life” perspectives. Prompts were developed from the original collection of notes to guide the model into reinterpreting the information to fit each character profile. 

    During this second stage the model expanded the notes into more detailed descriptions and adapted the tone to suit that of the character profile. Additionally, the model created the coherent “day in the life entries”, based on the human input. 

    In this way, the model acted as a generative tool rather than as an analysis tool, by taking the human selected and organised content and adapting it into the storytelling format. 

<h3>Part 2: Image Generation Using DALL-E</h3>

The images for this project were created using DALL-E, a generative AI system (OpenAI, 2022). DALL-E takes textual input and uses a diffusion model, which is a type of machine learning algorithm to refine information from the textual prompt into a coherent image (Naveed et al., 2025). 

These models are again trained on large datasets of image and text pairs, which allows them to learn the relationships between the visual output and the descriptive language or artistic style given in the initial input. Through using this algorithm, the generative AI can create the users desired image and align with their preferences in terms of style based on a text prompt. 

Similarly to the text generations, this section involved 2 main stages: 

-	The first stage involved creating textual prompts of the characters and scenes needed for the storytelling process. The details from these prompts included: 
o	The historical context of the travel journal
o	The type of character for each profile 
o	The location/ environment of each scene

The specific details in each prompt allowed the AI model to generate images that were easily related to each character profile and that were coherent with the story being told. 

-	The second stage involved refining the output from the generative AI to ensure visual consistency throughout the project and that the outputs created were still accurate to the scenes taken from the original travel journal. To do this, the textual prompts were iteratively refined to include further instructions on the style of the images, bearing in mind who the audience was and where the content was originally sourced from. 

Through repeated adjustments, the images generated began to align with the style of the project, and convey the original, often difficult to fully understand illustrations, into a more appropriate format. 

<h3>Part 3: Prompt Creation and Refinement</h3>

Human prompt development was a key part of working alongside AI tools within this project. Both the accuracy of the images and the character profiles relied on the details given in the initial input (Marvin et al., 2024). Rather than accepting the first generated output, the prompt development process was iterative and involved testing and then making further adjustments to ensure the final content generations were suitable for the project. This process highlighted that collaboration with generative AI tools requires continuous human input to create accurate desired content. 

Specifically for the text generation stages, prompts were developed for ChatGPT to guide the model into producing content that reflected the historical settings and social encounters expected withing 19th century travel. The initial prompts repeatedly generated overly dramatic storylines with language and references that were too modern. Due to this, the adjusted prompts provided further contextual information on language style and tone as well as additional emphasis on the time and events conveyed in the journal to ensure the profiles were accurate to the project. This process displayed the pattern recognition aspect of LLMs as the model relied heavily on the additional contextual to generate relevant text. 

The prompts for the character profiles were additionally designed with the individual themes for each perspective in mind. These included ideas surrounding the gendered perspectives, class motivations for travel and the comparisons between work and tourism. Through explicit mention of these themes in the prompts for each character, the generated profiles remained more consistent with the aim of the project to explore other perspectives of the journey through reference to the wider historical context described in the journal. Again, at this stage, human input remained essential to ensure the generated content did not fabricate historical details or contain inaccuracies. 

Below are examples of the prompts used in the text generation process: 

“Generate a short, historically accurate day in the life, styled diary entry of a 19th century, wealthy businessman travelling through Nice in 1867. Please make clear his freedoms as a male traveller, his class social status and the observations he makes. Also maintain a reflective tone of a 19th century traveller with considerations that the text will be for a non-specialist audience. Discuss topics like his accommodation, his travel by boat and rail and his daily activities.”

“Generate a short reflection from the perspective of the young modern male traveller on the experience of sightseeing in Nice” 

Prompt developing and refinement were equally important in the image generating process involving DALL-E. The prompts needed to include descriptors on specific time periods, clothing, environment and the overall style of the image. This was especially important when enhancing the original hand drawn images into more photorealistic versions. 

Below are examples of the prompts used for the image generation: 

“Generate a photorealistic image of a 19th century Victorian traveller, standing on the deck of a ship in 1867, wearing formal Victorian attire and carrying luggage. There should be overcast skies and be inspired by hand drawn illustrations with appropriate historical colours”

The process of prompt developing highlighted the strengths and weaknesses within the generative AI tools. Without the right descriptors, the text and images felt animated rather than the realistic style the project required. It again reinforced the importance of human input to endure that successful content was generated. 



**Visual Interpretation and Storytelling**

The final design stage of this project involved compiling the generated content into an accessible, storytelling format. After the “day in the life” styled diary entries, character profile images and scenario comparisons had all been created, they were then organised onto a series of digital profile cards. This format aimed to engage my younger audiences through the step-by-step process of getting to know the passengers, what their lives looked like and then their perspectives compared to modern comparisons of several key events depicted in the journal. 

To create the digital profile cards, Canva was used, as this platform allowed for the AI generated content to by styled beside the text in an easily engageable format (Canva, 2024). From the initial creation of each card, they were then organised into a logical order, starting with getting to know each of the characters and their diary entries and moving to the scenarios that they shared their commentary and comparisons about. This order aimed to help users to better contextualise the journals content. 


**Ethical Considerations and Limitations**

Ethical Considerations: 
Throughout the project there were ethical considerations as part working collaboratively with generative AI. These concerns centre around AI bias, interpreting historical material and the training data used for the generative AI models. Due to these ethical concerns, human oversight was necessary at every stage of the creative process to ensure that the generated content remained appropriate as an interpretation of the original journal and historically plausible. The project was also influences by the Lovelace-Boden-Ridler tests discussed in the coursework (Riedl, 2014; Boden, 2003; Ridler, 2021). The frameworks outlined in these theories helped to highlight areas that needed more careful consideration and how to best incorporate the generative AI models into the project while continuing to remain aware and informed of the ethical questions. 

Limitations: 
Below is the list of limitations encountered throughout the design and development of this project: 

-	The original journal was written as one perspective, a wealthy male traveller, the alternative profiles required further historical context, which was gained through use of additional sources, all referenced below. 
-	The AI outputs often used language that was too modern or in some cases too complex for my audience, careful refinement was carried out throughout the AI content generation process to ensure all text and images were important. 
-	The training datasets used by ChatGPT and DALL-E were not fully transparent to users, which created some uncertainty surrounding bias, however human oversight and research helped to combat this in the final outputs.
-	The generated images and character profiles should be taken as creative interpretations rather than as fully accurate historical reconstructions.
